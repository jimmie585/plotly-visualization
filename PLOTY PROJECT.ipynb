```python
import pandas  as pd 
import numpy as np
import plotly as py
import plotly.express as px
import seaborn as sns
import cufflinks as cf
cf.go_offline()
from plotly.offline import iplot
import plotly.io as pio
pio.renderers.default ='notebook'
py.offline.init_notebook_mode(connected= True)
```

```python
df=pd.DataFrame(np.random.rand(200,5), columns=['JAN','FEB','MAR','APR','MAY'])
df
```

```python
df.iplot()
```

```python
df.iplot(mode='markers')
```

```python
cf.datagen.sinwave(10,0.25).iplot()
```

```python
cf.datagen.sinwave(10,0.25).iplot(kind='surface',colorscale='accent')
```

```python
df=sns.load_dataset('tips')
df.head()
```

```python
import warnings
warnings.filterwarnings('ignore')
px.sunburst(df,path=['sex','smoker','day','time'],values='total_bill')
```

```python
px.pie(df,labels='day',values='total_bill',names='day')
```

```python
px.pie(df,labels='day',values='total_bill',names='day',hole= 0.6,title='A donut chart')
```

```python
df['total_bill'].iplot(kind='hist')
```

```python
px.histogram(df,y='tip')
```

```python
px.scatter(df,y='tip',x='total_bill')
```

```python
df.iplot(kind='scatter',y='tip',x='total_bill',mode='markers')
```
